# 01 — Explore one SMAP/MSL channel

Loads `P-1` when the NASA files are in `ml/data/`. The scaler is fit on the train file only. Windows use 250 steps of every column to predict the next value of column 0.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

import numpy as np
from src.config import WINDOW
from src.data import fit_scaler, load_channel, load_labels, transform, windows

try:
    train, test = load_channel('P-1')
except FileNotFoundError as exc:
    train = test = None
    print(exc)

if train is not None:
    labels = {row['chan_id']: row for row in load_labels()}
    print('train', train.shape, 'test', test.shape)
    print(labels['P-1'])
    scaler = fit_scaler(train)
    scaled = transform(scaler, train)
    inputs, targets = windows(scaled, WINDOW)
    print('windows', inputs.shape, 'targets', targets.shape, 'first target time', WINDOW)
    print('train col0 mean/std', float(train[:, 0].mean()), float(train[:, 0].std()))
```
